```python
def myFun(*args, **kwargs ):
    for arg in args:
        print(arg, end=" ")
    print("\n")

    print("Key Value Arguments (**kwargs) : \n")
    for key, value in kwargs.items():
        print(f"{key} = {value}")

myFun('Happy', 'Coding!', airoplain='Mode', front='Back', up='Down')
```
