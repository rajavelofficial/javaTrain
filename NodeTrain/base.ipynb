```python
def factorial(n):
    if n == 0:
        return 1
    else:
        return n * factorial(n - 1)
factorial(5)
```

```python
#Conditional Statement
num = int(input())

if(num > 0 ):
    print( num, "is a Positive number" )
elif(num < 0 ):
    print( num, "is Negative number")
else:
    print( num, "is Zero")
```
