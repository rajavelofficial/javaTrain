```python
def factorial(n):
    if n == 0:
        return 1
    else:
        return n * factorial(n - 1)
factorial(5)
```

```python
#Odd or Even Using if condition
num = int(input("Enter num : "))

if ( num % 2 == 0):
    print("Even")
else:
    print("Odd")
```
