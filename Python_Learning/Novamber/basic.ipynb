```python
def Argument(*args):
    for arg in args:
        print(arg)

Argument("Hello","welcome", "to", "Python")
```

```python
def keyValue(**kwargs):
    for key, value in kwargs.items():
        print(key+ ":" +value)

keyValue(Name = "Raja", Position = "Software Engineer", Experience = "2 Years")
```
